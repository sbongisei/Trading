```python
import numpy as np
```

log returns

```python
portfolio = [100, 120, 100, 80, 155]
log_returns = []
for i in range(len(portfolio)-1):
    log_returns.append(np.log(portfolio[i+1]/portfolio[i]))
print(log_returns)
```

```python
100 * np.exp(np.sum(log_returns)) == 155
```

Cumulative elog returns exercise

```python
portfolio = [100, 120, 100, 80, 155]
cum_log_returnas = []
cumulative = 0.0
for i in range(0,len(portfolio)):
    cumulative += np.log(portfolio[i+1]/portfolio[i])
    cum_log_returnas.append(cumulative)
```
